## Bivariate Analysis

```python
plt.figure(figsize=(10,5))
sns.barplot(data=Eds, x='category', y='Sales')
plt.xticks(rotation=45)
plt.title("Category vs Sales")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=Eds, x="category", y="profit")
plt.xticks(rotation=45)
plt.title("Category vs Profit")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=Eds, x='quantity', y='profit')
plt.xticks(rotation=45)
plt.title("quantity vs Profit")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=Eds, x='Discount', y='Sales')
plt.title("Discount vs Sales")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x="shipping_method",
y="delivery_days",data=Eds)
plt.xticks(rotation=45)
plt.title("Shipping Method vs Delivery Days")
plt.show()
```

```python
# correlation Metrix
plt.figure(figsize=(14,10))
sns.heatmap(Eds.select_dtypes(include="number").corr(),
            annot=False,
            cmap="coolwarm",
            linewidths=0.5,
           center=0)
plt.title("Correlation Heatmap",fontsize=16)
plt.tight_layout()
plt.show()
```
